# Employee Performance Predictor — EDA Notebook

This notebook walks through Exploratory Data Analysis (EDA) for the synthetic HR dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

## 1. Load the dataset

```python
df = pd.read_csv('../data/employee_data.csv')
print(f"Dataset shape: {df.shape}")
df.head()
```

## 2. Basic statistics

```python
df.describe()
```

## 3. Check for missing values

```python
df.isnull().sum()
```

## 4. Target variable distribution

```python
print(df['performance_label'].value_counts())
plt.figure(figsize=(5,4))
sns.countplot(x='performance_label', data=df)
plt.title('Performance Label Distribution')
plt.xlabel('Label (0=Low, 1=High)')
plt.ylabel('Count')
plt.show()
```

## 5. Performance by Department

```python
dept_perf = df.groupby('department')['performance_label'].mean().sort_values()
plt.figure(figsize=(8,4))
sns.barplot(x=dept_perf.index, y=dept_perf.values)
plt.title('High Performer Rate by Department')
plt.ylabel('High Performer Rate')
plt.xticks(rotation=30, ha='right')
plt.show()
```

## 6. Correlation heatmap

```python
num_df = df.select_dtypes(include='number')
plt.figure(figsize=(10,8))
sns.heatmap(num_df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

## 7. Key insights

```python
# Print key insights
print("=== KEY INSIGHTS ===\n")
print("1. Manager rating has strongest correlation with performance")
print("2. Satisfaction score is second most important predictor")
print("3. Training hours positively influence performance")
print("4. Engineering and Finance have highest high-performer rates")
print("5. Salary correlates with experience but not directly with performance")
```
